# First Look

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/results.csv")
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df.dtypes
```

```python
df.plot(x="created_at", y="download", kind="scatter", s=1, ylim=(0,1000), title="Download progression over time", ylabel="Download in Mbit/s", figsize=(20,10), alpha=0.5)
```

# Trend

```python
daily = df.groupby(df["created_at"].dt.date)["download"].mean()
daily
```

```python
daily.plot(ylim=(0,1000),figsize=(20,10), title="Average speeds over time in Mbit/s", ylabel="Download in Mbit/s")
```

```python
weekly = daily.rolling(7).mean()
weekly.plot(ylim=(0,1000),figsize=(20,10), title="7 day rolling average over time in Mbit/s", ylabel="Download in Mbit/s")
```

# Heatmap

```python
df["hour"] = df["created_at"].dt.hour
df["weekday"] = df["created_at"].dt.dayofweek
df.head()
```

```python
grid = df.pivot_table(index="weekday", columns="hour", values="download", aggfunc="mean")
grid
```

```python
plt.figure(figsize=(20, 10))
plt.title("Average download speed (Mbit/s) by weekday and hour")
sns.heatmap(grid,annot=True, fmt=".0f", cmap="RdYlGn", yticklabels=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
```
